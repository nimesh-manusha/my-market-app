```python
import pandas as pd
import numpy as np
from datetime import datetime
import os

os.makedirs('../data', exist_ok=True)
os.makedirs('../processing', exist_ok=True)

# Use the correct file name
df = pd.read_csv('../synthetic_online_retail_data.csv')

df.info()
df.head()

# Use the correct date column name
df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')

print("Missing values before cleaning:")
print(df.isnull().sum())

df = df.dropna()

# Calculate price per item (since there is no 'Sale Price' or 'Bottle Volume (ml)')
df['price_per_item'] = df['price'] / df['quantity']

df['Month'] = df['order_date'].dt.month
df['Day'] = df['order_date'].dt.day
df['Weekday'] = df['order_date'].dt.weekday
df['Year'] = df['order_date'].dt.year

# Save cleaned data
df.to_csv('../data/cleaned_data.csv', index=False)
print("Data cleaning completed. Cleaned data saved")
```
